# Gemini API Configuration

**Purpose:** Establish connection to Google's Gemini AI for retrieval-augmented generation capabilities.

**Business Function:** Enables AI-enhanced analysis that generates strategic business recommendations from customer data.

````python
import re
def _is_quota_exhausted(err: Exception) -> bool:
    """
    Best-effort detection for 'quota / credits exhausted' style failures.
    Gemini SDK exceptions vary a bit by version, so we check multiple signals.
    """
    msg = str(err).lower()

    # Common signals in the error text
    if "resource_exhausted" in msg:
        return True
    if re.search(r"\b429\b", msg):
        return True
    if "quota" in msg and ("exceed" in msg or "exhaust" in msg or "insufficient" in msg):
        return True

    # Sometimes the exception has a status_code attribute
    status_code = getattr(err, "status_code", None)
    if status_code == 429:
        return True

    return False
````

## Step 1.3: Vector Database Implementation

**Business Function:** Convert customer reviews into searchable vector representations for semantic retrieval.

**Technical Process:**
- Initialize sentence transformer model for text embedding generation
- Create ChromaDB collection for vector storage
- Transform customer reviews into 384-dimensional vector space
- Enable similarity-based search capabilities

**Action Required:** Execute without modification.

````python
# Initialize embedding model
print("Initializing sentence transformer model...")
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
print(f"Model configured: {embedding_model.get_sentence_embedding_dimension()}-dimensional vector space\n")

# Configure vector database
print("Configuring ChromaDB vector database...")
chroma_client = chromadb.Client()

# Remove existing collection if present
try:
    chroma_client.delete_collection(name="mobtel_reviews")
except:
    pass

# Create new collection
collection = chroma_client.create_collection(
    name="mobtel_reviews",
    metadata={"description": "MobTel customer review embeddings"}
)
print("Vector database initialized\n")

# Process and store customer reviews
print("Processing customer reviews for vector storage...")
documents = []
metadatas = []
ids = []

for idx, row in reviews_df.iterrows():
    # Filter for substantive reviews
    if pd.notna(row['review_text']) and len(str(row['review_text']).strip()) > 10:
        documents.append(str(row['review_text']))
        metadatas.append({
            'type': 'review',
            'product': str(row['product_model']),
            'rating': float(row['rating']),
            'date': str(row['review_date'])
        })
        ids.append(f"review_{idx}")

# Store in vector database
collection.add(
    documents=documents,
    metadatas=metadatas,
    ids=ids
)

print(f"Vector database populated: {len(documents)} customer reviews stored")
print("Semantic search system operational")
````

## Step 2.2: Cosine Similarity Calculation

**Business Function:** Quantify alignment between marketing descriptions and customer experiences.

**Technical Process:**
- Aggregate all customer reviews per product
- Generate embeddings for both marketing descriptions and customer reviews
- Calculate cosine similarity scores (0.0 to 1.0 scale)
- Classify alignment levels

**Similarity Score Interpretation:**
- **Strong Alignment (0.8-1.0)**: Marketing accurately reflects customer experience
- **Moderate Alignment (0.5-0.8)**: Some perception gaps exist, monitoring recommended
- **Weak Alignment (0.0-0.5)**: Significant marketing-reality mismatch, intervention required

**Action Required:** Execute without modification.

````python
print("Calculating marketing-experience alignment scores...\n")

# Aggregate customer reviews by product
product_reviews = reviews_df.groupby('product_model')['review_text'].apply(
    lambda x: ' '.join(str(v) for v in x if pd.notna(v))
).to_dict()

# Calculate similarity scores
similarity_results = []

for product in descriptions.keys():
    if product in product_reviews:
        # Retrieve description and aggregated reviews
        desc = descriptions[product]
        reviews = product_reviews[product]

        # Generate embeddings
        desc_embedding = embedding_model.encode([desc])
        review_embedding = embedding_model.encode([reviews])

        # Calculate cosine similarity
        similarity = cosine_similarity(desc_embedding, review_embedding)[0][0]

        # Classify alignment level
        if similarity >= 0.8:
            alignment = "Strong"
        elif similarity >= 0.5:
            alignment = "Moderate"
        else:
            alignment = "Weak"

        similarity_results.append({
            'Product': product,
            'Similarity Score': similarity,
            'Alignment Level': alignment
        })

# Generate results dataframe
similarity_df = pd.DataFrame(similarity_results).sort_values('Similarity Score', ascending=False)

print("Alignment analysis complete\n")
print("="*80)
print("\nMARKETING-EXPERIENCE ALIGNMENT ANALYSIS\n")
print(similarity_df.to_string(index=False))
print("\n" + "="*80)
````

**Action Required:** Execute without modification.

## Step 3.1: RAG System Implementation

**Business Function:** Combine document retrieval with AI reasoning to generate strategic business insights.

**System Architecture:**
1. Query vector database for relevant customer feedback
2. Provide retrieved context to Gemini AI
3. Generate evidence-based strategic recommendations
4. Return structured analysis with supporting data

**Action Required:** Execute without modification.

````python
def rag_query(business_question, n_retrieve=5):
    """
    Retrieval-Augmented Generation System

    Process:
    1. Execute semantic search for relevant customer feedback
    2. Construct business analysis prompt with retrieved context
    3. Generate AI-enhanced strategic recommendations
    4. Return structured business insights
    """

    print(f"\nStep 1: Retrieving relevant customer feedback...")

    # Execute vector search
    results = collection.query(
        query_texts=[business_question],
        n_results=n_retrieve
    )

    # Aggregate retrieved documents
    context = "\n\n".join(results['documents'][0])
    print(f"Retrieved {len(results['documents'][0])} relevant customer reviews\n")

    print(f"Step 2: Generating AI-enhanced strategic analysis...")

    # Construct business analysis prompt
    prompt = f"""
You are a senior business strategy consultant analyzing customer feedback for MobTel Communications, a mobile phone retailer.

BUSINESS QUESTION:
{business_question}

CUSTOMER FEEDBACK DATA:
{context}

ANALYSIS REQUIREMENTS:
1. Identify key patterns and themes in customer feedback
2. Assess business implications and strategic opportunities
3. Generate 3-5 specific, actionable recommendations
4. For each recommendation, provide:
   - Specific action to be taken
   - Supporting evidence from customer data
   - Expected business impact (revenue, satisfaction, efficiency)
   - Implementation timeline and stakeholder ownership

FORMAT REQUIREMENTS:
## Executive Summary
[2-3 sentence overview of key findings]

## Analysis of Customer Feedback Patterns
[Identify what the data reveals about customer sentiment, product perception, and business performance]

## Strategic Recommendations

### Recommendation 1: [Clear, specific title]
**Action:** [Precise description of what MobTel should do]
**Evidence:** [Specific customer feedback supporting this recommendation]
**Business Impact:** [Quantified or qualified expected outcome]
**Implementation:** [Timeline and responsible stakeholder]

[Repeat format for recommendations 2-5]

## Risk Assessment
[Identify risks of not acting on these recommendations]
"""

    # Generate AI insights
    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=[prompt]
    )

    print(f"AI analysis complete\n")

    return {
        'question': business_question,
        'retrieved_docs': results['documents'][0],
        'ai_insights': response.text,
        'num_sources': len(results['documents'][0])
    }

print("RAG system configured successfully")
````
