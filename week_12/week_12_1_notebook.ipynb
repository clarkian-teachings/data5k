# **ChromaDB: Vector Search vs. RAG (Retrieval-Augmented Generation)**

This notebook demonstrates the difference between **Vector Search** and **RAG (Retrieval-Augmented Generation)** using ChromaDB. You'll see how to query the vector database directly and also how to combine ChromaDB with an LLM to generate contextual answers.

### **1 Set Up ChromaDB with Vector Embeddings**

```python
import tempfile
'''
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import Chroma
from langchain.schema import Document
'''


from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document





# Step 1: Create a temporary directory for ChromaDB storage
temp_dir = tempfile.mkdtemp()
print(f"📂 Temporary directory created at: {temp_dir}")

# Step 2: Initialize the HuggingFace embedding model
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Step 3: Create a Chroma vector store with persistence in the temp directory
chroma_db = Chroma(
    collection_name="example_docs",
    embedding_function=embedding_model,
    persist_directory=temp_dir
)

# Step 4: Add sample documents to the ChromaDB
documents = [
    "Machine learning enables machines to learn from data.",
    "Deep learning uses artificial neural networks to perform learning tasks.",
    "ChromaDB allows metadata support for vector storage.",
    "HuggingFace models power NLP applications.",
    "Natural language processing (NLP) allows computers to understand human language.",
    "Python is a versatile programming language widely used in AI and machine learning.",
    "LangChain allows developers to chain together different AI models and tools."
]

# Convert plain text into Document objects
document_objects = [Document(page_content=doc) for doc in documents]

# Add documents to the ChromaDB vector store
chroma_db.add_documents(document_objects)

# Persist the vector store to the temp directory
chroma_db.persist()

print(f"✅ ChromaDB setup complete. The vector store has been saved to {temp_dir}")
```

### **2. Query ChromaDB (Vector Search Only, No LLM)**

```python

# Reload the ChromaDB vector store from the temp directory
chroma_db = Chroma(
    collection_name="example_docs",
    embedding_function=embedding_model,
    persist_directory=temp_dir
)

# Query the vector database
query = "What is machine learning?"
results = chroma_db.similarity_search(query, k=3)

# Display the results
print(f"🔍 Query: {query}\n")
for i, result in enumerate(results):
    print(f"Result {i+1}:")
    print(f"Document: {result.page_content}\n")
```



### **3 RAG (Retrieval-Augmented Generation) with LLM**

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

# Step 1: Initialize GPT-Neo (1.3B model)
gpt_neo_model_name = "EleutherAI/gpt-neo-1.3B"  # You can change this to 2.7B or 125M if needed
tokenizer = AutoTokenizer.from_pretrained(gpt_neo_model_name)
model = AutoModelForCausalLM.from_pretrained(gpt_neo_model_name)

# Step 2: Retrieve top-3 documents from ChromaDB
query = "Explain deep learning in simple terms."
results = chroma_db.similarity_search(query, k=3)

# Step 3: Combine the top-3 results into a context for GPT-Neo
context = "\n\n".join([result.page_content for result in results])

# Step 4: Use GPT-Neo to generate an answer based on the retrieved documents
prompt = f"""
You are an AI assistant. Use the following context to answer the user's question.

Context:
{context}

Question: {query}
Answer in a simple and concise way:
"""

# Tokenize the prompt for GPT-Neo
inputs = tokenizer(prompt, return_tensors="pt")

# Generate a response using GPT-Neo
output = model.generate(
    inputs['input_ids'],
    max_length=150,  # Adjust the max_length as needed
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7  # Control randomness in the response
)

# Decode the output and display the response
response = tokenizer.decode(output[0], skip_special_tokens=True)
print(f"🤖 **GPT-Neo Response**:")
print(response)
```

### **Summary of Key Differences**


| **Task**          | **Vector Search (Chroma Only)** | **RAG (Chroma + LLM)**         |
|------------------|---------------------------------|---------------------------------|
| **Find Documents** | ✅ Yes, finds similar docs    | ✅ Yes, finds similar docs    |
| **Synthesize Answer** | ❌ No reasoning            | ✅ LLM generates logical answer |
| **Needs LLM?**    | ❌ No                          | ✅ Yes                         |
| **Speed**         | ⚡ Faster                     | 🐢 Slower                      |
| **Use Case**      | Document lookup only         | Generate contextual answers     |

**Key Takeaways:**
- **Vector Search**: Fast, mathematical matching of documents, but no reasoning.
- **RAG (Retrieval-Augmented Generation)**: Uses LLMs to generate synthesized answers from retrieved documents.
