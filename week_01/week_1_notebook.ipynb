**A lot of models will require the Date column to be a format of date/datetime**

Pandas provides many methods to change the datatype. In the example below we change the column Date to be a DateTime.

```python
df["AU-CPI"] = df["AU-CPI"].astype("int64") # Try and change the datatype of the column AU-CPI to be integer

# An integer is a whole number with no fractional part (… -2, -1, 0, 1, 2 …).
# A float (“floating-point number”) can represent numbers with a fractional part (… 1.5, 2.0, 3.14159 …)
```

```python
df.head() # Now double check that the change was successful

```

```python
df["AU-CPI"] = df["AU-CPI"].astype("float64") # Lastly change the datatype back to its original float type
```
